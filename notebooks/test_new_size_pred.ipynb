```python
dataset_path = "/home/zakarianarjis/Downloads/brightfield_particles.hdf5"
classes = [2,3]
plot_dataset = ParticleDatasetReg(
    h5_path=dataset_path,
    classes=classes,
    mean=None,
    std=None,
    padding=False,
    transform=None,
    indices=None,
)
```

```python
import torch
from torch import nn
device = "cuda:0" if torch.cuda.is_available() else "cpu"
model = ResNet18(num_classes=1)  # For regression, output is a single value
checkpoint_path = "/home/zakarianarjis/workspace/iscat/experiments/runs/size_prediction/ResNet18_Regression_2025-07-03_15-45-53/best_model.pth"
if os.path.exists(checkpoint_path):
    model.load_state_dict(torch.load(checkpoint_path, map_location=device,weights_only=False)['model_state_dict'])
model = model.to(device)
```

```python
dataloader = DataLoader(
    plot_dataset,
    batch_size=len(plot_dataset),
    shuffle=False,
    num_workers=12,
    pin_memory=True,
)
for batch in dataloader:
    images, labels,size_label = batch
    images = images.to(device)
    size_label = labels.to(device)

    with torch.no_grad():
        outputs = model(images)
        outputs = outputs.squeeze(1)  # Remove the extra dimension for regression

    # print("Labels:", size_label)
    # print("Outputs:", outputs)
    # print("Mean Absolute Error:", nn.functional.l1_loss(outputs, size_label).item())
    # print("Mean Squared Error:", nn.functional.mse_loss(outputs, size_label).item())
```
